```python
#!/usr/bin/env python3
"""
TdP v2.4 — THE MINIMAL CORE
A Research Protocol for G₂ Topological Phenomenology

Author: Carlos Herrero González | February 2026
License: MIT

PHILOSOPHY
----------
This script implements a minimal research protocol, not a proven theory.
It contains exactly two elements:

1. PREDICTION (falsifiable):
   δ_CP = 197.8° derived purely from G₂ topology (b₂=13, b₃=7).
   → Testable at DUNE (2030). Exclusion criterion: |δ_CP - 197.8°| > 10° at 3σ.

2. CLUE (phenomenological):
   Lepton masses scale as φ^[0, 11, 17] with <4% error.
   → 11 = M-theory dimension; 6 = b₂-b₃ (G₂ moduli space); 17 = 11+6.
   → NOT claimed as law. Labeled explicitly as an intriguing coincidence
      requiring rigorous derivation from the Dirac operator on G₂.

STRATEGY
--------
• No overclaiming: All speculative elements carry explicit methodological warnings.
• No fine-tuning: Only m_e calibrates the mass scale; μ and τ are predictions.
• No cherry-picking: No formulas for α, M_Z, M_H without geometric mechanism.
• Falsifiability first: The entire protocol stands or falls on DUNE's δ_CP.

This is a protocol for empirical validation—not a theory demanding belief.
"""

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.patches import Rectangle, FancyBboxPatch
import matplotlib.patches as mpatches

# ── 1. FIXED TOPOLOGICAL INVARIANTS (G₂ TCS MANIFOLD) ────────────────────────
B3 = 7                          # Associative cycles (Betti b₃)
B2 = 13                         # Coassociative cycles (Betti b₂)
K_CS = B3 * B2                  # Chern-Simons level = 91
PHI = (1 + np.sqrt(5)) / 2      # Golden ratio (intrinsic to G₂ holonomy)

# ── 2. PILLAR #1: δ_CP PREDICTION (PURE TOPOLOGY) ────────────────────────────
def delta_cp_prediction(b3: int = B3, k_cs: int = K_CS) -> float:
    """
    Predict δ_CP phase from pure G₂ topology.

    Formula: δ_CP = 2π × (b₃² + 1) / k_CS

    This is a parameter-free prediction derived solely from integer
    topological invariants. No fitting to experimental data.

    Returns
    -------
    float
        Predicted δ_CP in degrees (197.80°)
    """
    return np.degrees(2 * np.pi * (b3**2 + 1) / k_cs)

DELTA_CP_PRED = delta_cp_prediction()

# Current experimental constraints (NuFit 5.2 global fit, 2024)
# Source: JHEP 01 (2023) 071 — NuFit 5.2 (including Super-K, T2K, NOvA)
NUFIT_DELTA_CP_BF = 197.0       # Best fit
NUFIT_DELTA_CP_1SIGMA = (175.0, 238.0)   # 1σ range
NUFIT_DELTA_CP_3SIGMA = (142.0, 340.0)   # 3σ range

# ── 3. PILLAR #2: LEPTON MASS SCALING (PHENOMENOLOGICAL CLUE) ────────────────
# Exponents: [0, 11, 17] where:
#   • 11 = dimension of M-theory (11D supergravity)
#   • 6  = B2 - B3 = dimension of G₂ moduli space
#   • 17 = 11 + 6 (topological sum)
MASS_EXPONENTS = np.array([0, 11, 17])
LEPTON_NAMES = ["e⁻", "μ⁻", "τ⁻"]
MASS_EXP_MEV = np.array([0.510998950, 105.6583755, 1776.86])  # PDG 2024

# Calibrate scale exclusively on m_e (g=1) — no fitting to μ or τ
V_H = MASS_EXP_MEV[0]  # Electron mass sets the scale

def lepton_mass(g_index: int, phi: float = PHI, v_h: float = V_H) -> float:
    """
    Unified lepton mass function based on topological exponents.

    M(g) = V_H × φ^exponent[g]

    Parameters
    ----------
    g_index : int
        Generation index (0=e, 1=μ, 2=τ)
    phi : float
        Golden ratio (geometric eigenvalue candidate)
    v_h : float
        Mass scale calibrated exclusively on m_e

    Returns
    -------
    float
        Predicted lepton mass in MeV

    Note
    ----
    This is a phenomenological clue, NOT a derived law. The exponents
    [0,11,17] reproduce data within <4% but lack rigorous derivation from
    the Dirac operator spectrum on G₂ manifolds.
    """
    return v_h * phi ** MASS_EXPONENTS[g_index]

MASS_PRED_MEV = np.array([lepton_mass(i) for i in range(3)])
RELATIVE_ERROR = 100 * (MASS_PRED_MEV - MASS_EXP_MEV) / MASS_EXP_MEV

# ── 4. VISUALIZATION MODULE (DARK MODE SCIENTIFIC STYLE) ─────────────────────
def configure_dark_style():
    """Configure matplotlib for dark-mode scientific aesthetic."""
    sns.set_style("darkgrid", {
        'axes.facecolor': '#0f0f15',
        'figure.facecolor': '#0a0a10',
        'grid.color': '#2a2a35',
        'text.color': '#e0e0ff',
        'axes.labelcolor': '#c0c0e0',
        'xtick.color': '#b0b0d0',
        'ytick.color': '#b0b0d0',
        'axes.edgecolor': '#404055'
    })
    plt.rcParams.update({
        'font.family': 'DejaVu Sans',
        'font.size': 10,
        'axes.titlesize': 13,
        'axes.labelsize': 11,
        'legend.fontsize': 9,
        'figure.titlesize': 14
    })

def plot_dune_landscape():
    """Visualize δ_CP prediction against current experimental constraints."""
    fig, ax = plt.subplots(figsize=(8, 4.5))

    # 3σ band (NuFit)
    ax.axvspan(NUFIT_DELTA_CP_3SIGMA[0], NUFIT_DELTA_CP_3SIGMA[1],
               color='#1a3a5c', alpha=0.4, label='NuFit 5.2 (3σ)')
    # 1σ band (NuFit)
    ax.axvspan(NUFIT_DELTA_CP_1SIGMA[0], NUFIT_DELTA_CP_1SIGMA[1],
               color='#2a5a8c', alpha=0.6, label='NuFit 5.2 (1σ)')
    # Best fit
    ax.axvline(NUFIT_DELTA_CP_BF, color='#64b5f6', linestyle='--',
               linewidth=1.8, label=f'NuFit BF: {NUFIT_DELTA_CP_BF}°')

    # TdP prediction
    ax.axvline(DELTA_CP_PRED, color='#ff6b6b', linewidth=2.5,
               label=f'TdP Prediction: {DELTA_CP_PRED:.1f}°', zorder=5)

    # Confirmation zone (±5° around prediction)
    confirmation_zone = Rectangle((DELTA_CP_PRED - 5, 0), 10, 1.2,
                                  facecolor='#4caf50', alpha=0.3,
                                  label='Confirmation Zone (±5°)')
    ax.add_patch(confirmation_zone)

    # Exclusion boundary (±10°)
    ax.axvline(DELTA_CP_PRED - 10, color='#ff9800', linestyle=':',
               linewidth=1.5, alpha=0.7)
    ax.axvline(DELTA_CP_PRED + 10, color='#ff9800', linestyle=':',
               linewidth=1.5, alpha=0.7)

    ax.set_xlim(120, 360)
    ax.set_ylim(0, 1)
    ax.set_xlabel('δ_CP (degrees)', fontsize=12, color='#e0e0ff')
    ax.set_title('DUNE Prediction Landscape: δ_CP = 197.8°',
                 fontsize=14, color='#ffffff', pad=15, weight='bold')
    ax.legend(loc='upper right', framealpha=0.92,
              facecolor='#151525', edgecolor='#404055')
    ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.7)

    # Annotation
    ax.text(DELTA_CP_PRED, 1.05, 'TdP v2.4',
            ha='center', color='#ff6b6b', fontsize=11, weight='bold')
    ax.text(DELTA_CP_PRED, 0.92, 'Falsifiable at DUNE (2030)\nExclusion: |Δ| > 10° @ 3σ',
            ha='center', color='#aaaaaa', fontsize=8.5)

    plt.tight_layout()
    return fig

def plot_lepton_scaling():
    """Visualize lepton mass scaling law on semi-log plot."""
    fig, ax = plt.subplots(figsize=(8, 4.5))

    # Theoretical φ^n curve (continuous)
    n_fine = np.linspace(0, 18, 200)
    mass_curve = V_H * PHI ** n_fine
    ax.plot(n_fine, mass_curve, color='#8a65cc', linestyle='--',
            linewidth=1.8, alpha=0.7, label=f'φⁿ scaling (V_H = {V_H:.3f} MeV)')

    # Experimental points
    ax.scatter(MASS_EXPONENTS, MASS_EXP_MEV,
               s=180, c='#ff6b6b', marker='o', edgecolors='#ffffff',
               linewidths=1.8, zorder=5, label='Experimental masses (PDG 2024)')

    # Predicted points (should align with curve)
    ax.scatter(MASS_EXPONENTS, MASS_PRED_MEV,
               s=120, c='#4caf50', marker='x', linewidths=2.5,
               zorder=6, label='TdP prediction (calibrated on m_e only)')

    # Annotations for each lepton
    for i, (n, m_exp, m_pred) in enumerate(zip(MASS_EXPONENTS, MASS_EXP_MEV, MASS_PRED_MEV)):
        offset = 15 if i == 2 else -25
        ax.annotate(f'{LEPTON_NAMES[i]}\nexp: {m_exp:.1f} MeV\npred: {m_pred:.1f} MeV',
                    xy=(n, m_exp), xytext=(0, offset), textcoords='offset points',
                    ha='center', fontsize=9, color='#e0e0ff',
                    bbox=dict(boxstyle='round,pad=0.4', facecolor='#1a1a2a',
                              edgecolor='#404055', alpha=0.85))

    ax.set_yscale('log')
    ax.set_xlabel('Topological Exponent n ∈ {0, 11, 17}',
                  fontsize=12, color='#e0e0ff')
    ax.set_ylabel('Lepton Mass (MeV) — Log Scale',
                  fontsize=12, color='#e0e0ff')
    ax.set_title('Lepton Scaling Law: M ∝ φⁿ with n = {0, 11, 17}',
                 fontsize=14, color='#ffffff', pad=15, weight='bold')
    ax.set_xticks([0, 5, 10, 11, 15, 17])
    ax.set_xticklabels(['0', '5', '10', '11\n(M-theory)', '15', '17\n(11+6)'])
    ax.legend(loc='upper left', framealpha=0.92,
              facecolor='#151525', edgecolor='#404055')
    ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.7)

    # Geometric interpretation annotation
    ax.text(0.02, 0.97,
            'Geometric interpretation:\n'
            '• n=11 → M-theory dimension (11D supergravity)\n'
            '• n=6  → G₂ moduli space dim (b₂-b₃=6)\n'
            '• n=17 → 11 + 6 (topological sum)',
            transform=ax.transAxes, fontsize=8.5, color='#a0a0c0',
            verticalalignment='top',
            bbox=dict(boxstyle='round,pad=0.6', facecolor='#0f1525',
                      edgecolor='#3a3a55', alpha=0.8))

    plt.tight_layout()
    return fig

def plot_error_analysis():
    """Visualize relative error without fine-tuning."""
    fig, ax = plt.subplots(figsize=(8, 3.8))

    colors = ['#4caf50' if abs(e) < 4 else '#ff9800' for e in RELATIVE_ERROR]
    bars = ax.bar(LEPTON_NAMES, RELATIVE_ERROR,
                  color=colors, edgecolor='#ffffff', linewidth=1.5, alpha=0.85)

    # Zero line
    ax.axhline(0, color='#606075', linestyle='-', linewidth=1.2)

    # ±4% tolerance band (no fine-tuning threshold)
    ax.axhspan(-4, 4, color='#2a4a2a', alpha=0.25, label='No fine-tuning zone (|ε| < 4%)')

    # Annotations
    for i, (bar, err) in enumerate(zip(bars, RELATIVE_ERROR)):
        height = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2., height + (2.5 if height > 0 else -3.5),
                f'{err:+.2f}%', ha='center', va='bottom' if height > 0 else 'top',
                fontsize=10, weight='bold', color='#e0e0ff')

    ax.set_ylabel('Relative Error ε (%)', fontsize=12, color='#e0e0ff')
    ax.set_title('Relative Error Analysis: No Fine-Tuning Required',
                 fontsize=14, color='#ffffff', pad=15, weight='bold')
    ax.set_ylim(-5, 5)
    ax.legend(loc='upper right', framealpha=0.92,
              facecolor='#151525', edgecolor='#404055')
    ax.grid(True, axis='y', alpha=0.3, linestyle='--', linewidth=0.7)

    # Methodology footnote
    ax.text(0.5, -0.22,
            'Methodology: Mass scale V_H calibrated exclusively on m_e (g=1).\n'
            'μ and τ masses are predictions—no parameters adjusted to fit data.',
            transform=ax.transAxes, ha='center', fontsize=8.5, color='#a0a0c0')

    plt.tight_layout()
    return fig

def generate_protocol_dashboard():
    """Generate the complete TdP v2.4 protocol dashboard."""
    configure_dark_style()

    print("=" * 80)
    print("TdP v2.4 — THE MINIMAL CORE (Research Protocol)")
    print("=" * 80)
    print("\n🔬 PILLAR #1: δ_CP PREDICTION (Pure Topology)")
    print(f"   Formula: δ_CP = 2π × (b₃² + 1) / k_CS")
    print(f"   Input:   b₃ = {B3}, b₂ = {B2}, k_CS = {K_CS}")
    print(f"   Output:  δ_CP = {DELTA_CP_PRED:.2f}°")
    print(f"   Status:  Falsifiable at DUNE (2030)")
    print(f"   Exclusion criterion: |δ_CP^DUNE - 197.8°| > 10° at 3σ")

    print("\n🔍 PILLAR #2: LEPTON MASS SCALING (Phenomenological Clue)")
    print(f"   Scaling law: M(g) = V_H × φ^[0, 11, 17]")
    print(f"   Calibration: V_H set exclusively by m_e = {V_H:.6f} MeV")
    print(f"   Geometric hint: 11 = dim(M-theory), 6 = b₂-b₃, 17 = 11+6")
    print(f"   Status: Intriguing coincidence (<4% error) — NOT a derived law")
    print(f"   Methodological warning: No rigorous derivation from Dirac operator on G₂")

    print("\n📊 Generating visualization dashboard...")
    fig1 = plot_dune_landscape()
    fig2 = plot_lepton_scaling()
    fig3 = plot_error_analysis()

    print("\n✅ Dashboard generated successfully.")
    print("\n" + "=" * 80)
    print("PROTOCOL PHILOSOPHY")
    print("=" * 80)
    print("""
1. PREDICTION FIRST:
   The only claim requiring belief is δ_CP = 197.8° — a parameter-free
   prediction testable within 4 years. Everything else is provisional.

2. CLUES SECOND:
   The [0,11,17] mass scaling is presented transparently as an unexplained
   coincidence. It gains significance ONLY if DUNE confirms δ_CP.

3. NO OVERCLAIMING:
   No formulas for α, M_Z, or Higgs without geometric mechanism.
   No assertions about "decoding the universe."

4. FALSIFIABILITY AS INTEGRITY:
   This protocol can be killed cleanly by experiment. That is its strength.
   """)
    print("=" * 80)
    print("\n💡 This is a research protocol — not a theory demanding belief.")
    print("   Its value lies in empirical testability, not numerical elegance.\n")

    return fig1, fig2, fig3

# ── 5. EXECUTE PROTOCOL ──────────────────────────────────────────────────────
if __name__ == "__main__":
    figs = generate_protocol_dashboard()
    plt.show()
```
